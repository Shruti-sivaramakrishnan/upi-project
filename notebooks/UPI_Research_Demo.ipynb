```python
import pandas as pd

def find_header_row(filepath):
    """Scan the first few rows to find which one has 'UPI Remitter Banks'."""
    if filepath.endswith('.csv'):
        probe = pd.read_csv(filepath, header=None, nrows=5)
    else:
        probe = pd.read_excel(filepath, header=None, nrows=5)
    for i in range(len(probe)):
        if probe.iloc[i].astype(str).str.contains('UPI Remitter Banks').any():
            return i
    raise ValueError(f"Could not find header row in {filepath}")

def parse_td(val):
    """Handle both '0.32%' text and 0.0084 fraction formats."""
    if isinstance(val, str):
        return float(val.replace('%', ''))
    else:
        return float(val) * 100  # fraction -> percent

def clean_upi_month(filepath, month_label, target_banks):
    header_row = find_header_row(filepath)
    if filepath.endswith('.csv'):
        raw = pd.read_csv(filepath, header=header_row)
    else:
        raw = pd.read_excel(filepath, header=header_row)

    filtered = raw[raw['UPI Remitter Banks'].isin(target_banks)].copy()
    if filtered.shape[0] != len(target_banks):
        print(f"⚠️ WARNING: expected {len(target_banks)} banks, got {filtered.shape[0]} for {month_label}")
        print("Unique bank names found:", raw['UPI Remitter Banks'].unique())

    clean = filtered[['UPI Remitter Banks', 'Total Volume (In Mn)', 'TD%']].copy()
    clean.columns = ['bank', 'total_volume_mn', 'td_pct']
    clean['td_pct'] = clean['td_pct'].apply(parse_td)
    clean['month'] = month_label
    return clean
```

```python
import pandas as pd
import os

def standardize_columns(columns):
    """
    Convert column names to a standard format.
    """

    cleaned = []

    for col in columns:

        col = str(col)

        col = col.strip()

        col = col.lower()

        col = col.replace(" ", "_")

        cleaned.append(col)

    return cleaned


def read_monthly_file(file_path):
    """
    Reads one NPCI monthly file and returns
    a clean DataFrame with standardized columns.
    """

    extension = os.path.splitext(file_path)[1].lower()

    # Read raw file without assuming header
    if extension == ".xlsx":
        raw = pd.read_excel(file_path, header=None, dtype=str)

    elif extension == ".csv":
        raw = pd.read_csv(file_path, header=None, dtype=str)

    else:
        raise ValueError("Unsupported file type.")

    # Detect header row
    header_row = None

    for i in range(min(10, len(raw))):

        row_text = " ".join(
            raw.iloc[i]
            .fillna("")
            .astype(str)
            .str.lower()
        )

        if (
            "td" in row_text and
            "approved" in row_text and
            "remitter" in row_text
        ):
            header_row = i
            break

    if header_row is None:
        raise ValueError(f"Header not found in {file_path}")

    # Read again using detected header
    if extension == ".xlsx":
        df = pd.read_excel(
            file_path,
            header=header_row
        )

    else:
        df = pd.read_csv(
            file_path,
            header=header_row
        )

    # Standardize column names
    df.columns = standardize_columns(df.columns)

    return df
```

```python
import pandas as pd
import os
import glob
import re

# -----------------------------
# Target Banks
# -----------------------------
TARGET_BANKS = [
    "State Bank of India",
    "Bank of Baroda",
    "HDFC Bank Ltd.",
    "Union Bank of India",
    "Punjab National Bank"
]

# -----------------------------
# Standardize Column Names
# -----------------------------
def standardize_columns(columns):

    cleaned = []

    for col in columns:
        col = str(col).strip().lower()

        col = col.replace(" ", "_")
        col = col.replace(".", "")
        col = col.replace("%", "pct")
        col = col.replace("(", "")
        col = col.replace(")", "")

        cleaned.append(col)

    return cleaned


# -----------------------------
# Read Monthly File
# -----------------------------
def read_monthly_file(file_path):

    extension = os.path.splitext(file_path)[1].lower()

    if extension == ".xlsx":
        raw = pd.read_excel(file_path, header=None, dtype=str)

    elif extension == ".csv":
        raw = pd.read_csv(file_path, header=None, dtype=str)

    else:
        raise ValueError("Unsupported file type")

    header_row = None

    for i in range(min(10, len(raw))):

        row_text = " ".join(
            raw.iloc[i]
            .fillna("")
            .astype(str)
            .str.lower()
        )

        if (
            "approved" in row_text
            and "td" in row_text
            and ("remitter" in row_text)
        ):
            header_row = i
            break

    if header_row is None:
        raise ValueError(f"Header not found in {file_path}")

    if extension == ".xlsx":
        df = pd.read_excel(file_path, header=header_row)

    else:
        df = pd.read_csv(file_path, header=header_row)

    df.columns = standardize_columns(df.columns)

    # Make all files use the same schema
    df.rename(columns={
        "upi_remitter_members": "upi_remitter_banks",
        "upi_remitter_member": "upi_remitter_banks",
        "upi_remitter_bank": "upi_remitter_banks",
        "upi_remitterbanks": "upi_remitter_banks",
        "tdpct": "td_pct",
        "td": "td_pct"
    }, inplace=True)

    return df


# -----------------------------
# Extract Month from Filename
# -----------------------------
def extract_month(filename):

    months = {
        "Jan":"01",
        "Feb":"02",
        "Mar":"03",
        "Apr":"04",
        "May":"05",
        "Jun":"06",
        "Jul":"07",
        "Aug":"08",
        "Sep":"09",
        "Oct":"10",
        "Nov":"11",
        "Dec":"12"
    }

    match = re.search(
        r"(20\d{2})-(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)",
        filename
    )

    if match:
        return f"{match.group(1)}-{months[match.group(2)]}"

    return None


# -----------------------------
# Find Files
# -----------------------------
all_files = []

all_files.extend(glob.glob("/content/*Remitter.xlsx"))
all_files.extend(glob.glob("/content/*Remitter*.csv"))

all_files = sorted(all_files)

print(f"Found {len(all_files)} files\n")


# -----------------------------
# Build Master Dataset
# -----------------------------
master_df = pd.DataFrame()

for file in all_files:

    print("Processing:", os.path.basename(file))

    df = read_monthly_file(file)

    df = df[df["upi_remitter_banks"].isin(TARGET_BANKS)].copy()

    df["month"] = extract_month(os.path.basename(file))

    df = df[
        [
            "month",
            "upi_remitter_banks",
            "total_volume_in_mn",
            "td_pct"
        ]
    ]

    master_df = pd.concat(
        [master_df, df],
        ignore_index=True
    )


print("\nProcessing Complete!\n")

print("Master DataFrame Shape:")
print(master_df.shape)

print("\nRows per Month:")
print(master_df.groupby("month").size())

display(master_df.head(15))
```

```python
def normalize_td(value):
    """
    Convert all TD values to percentage (%).

    Examples:
        0.0084  -> 0.84
        "0.32%" -> 0.32
    """

    if pd.isna(value):
        return None

    value = str(value).strip()

    # Already stored as percentage string
    if value.endswith("%"):
        return float(value.replace("%", ""))

    # Stored as decimal fraction
    value = float(value)

    return value * 100


# Apply normalization
master_df["td_pct"] = master_df["td_pct"].apply(normalize_td)

print("TD column converted successfully!\n")

display(master_df.head(15))
```

```python
master_df["month"] = pd.to_datetime(master_df["month"])

master_df = master_df.sort_values(
    ["month", "upi_remitter_banks"]
).reset_index(drop=True)

display(master_df.head(15))
```

```python
# Calculate historical statistics for each bank
bank_stats = (
    master_df
    .groupby("upi_remitter_banks")["td_pct"]
    .agg(["mean", "std"])
    .reset_index()
)

print("Historical TD% statistics for each bank:")
display(bank_stats)
```

```python
# Add bank mean and standard deviation to every row
master_df = master_df.merge(
    bank_stats,
    on="upi_remitter_banks",
    how="left"
)

display(master_df.head(10))
```

```python
# Calculate z-score for each bank-month
master_df["z_score"] = (
    (master_df["td_pct"] - master_df["mean"]) /
    master_df["std"]
)

display(master_df.head(15))
```

```python
# ---------------------------------------
# Compute Monthly ISS
# ---------------------------------------

iss_monthly = (
    master_df
    .groupby("month")
    .agg(
        iss_mean_z=("z_score", "mean"),
        iss_max_z=("z_score", "max")
    )
    .reset_index()
)

# Placeholder threshold (Exploratory)
iss_monthly["stress_flag"] = iss_monthly["iss_mean_z"] > 1.5

print("Monthly ISS:")
display(iss_monthly)
```

```python
import pandas as pd
import numpy as np
import os

def clean_upi_month(file_path, month_label, target_banks):

    # -----------------------------
    # Read file
    # -----------------------------
    if file_path.lower().endswith(".csv"):
        raw = pd.read_csv(file_path, header=None, dtype=str)
    else:
        raw = pd.read_excel(file_path, header=None, dtype=str)

    # -----------------------------
    # Find header row
    # -----------------------------
    header_row = None

    for i in range(len(raw)):

        row = (
            raw.iloc[i]
            .fillna("")
            .astype(str)
            .str.lower()
            .tolist()
        )

        row_text = " ".join(row)

        if (
            "sr. no." in row_text
            and (
                "upi remitter banks" in row_text
                or "upi_remitter_banks" in row_text
                or "upi remitter members" in row_text
                or "upi_remitter_members" in row_text
            )
        ):
            header_row = i
            break

    if header_row is None:
        raise ValueError(
            f"Header row not found in {os.path.basename(file_path)}"
        )

    # -----------------------------
    # Read again using correct header
    # -----------------------------
    if file_path.lower().endswith(".csv"):
        df = pd.read_csv(file_path, header=header_row)
    else:
        df = pd.read_excel(file_path, header=header_row)

    # -----------------------------
    # Standardize column names initially
    # -----------------------------
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
    )

    # -----------------------------
    # Detect bank column and rename it
    # -----------------------------
    bank_col_name = None

    for col in df.columns:

        c = col.lower()

        if (
            "remitter" in c
            and (
                "bank" in c
                or "member" in c
            )
        ):
            bank_col_name = col
            break

    if bank_col_name is None:
        raise ValueError("Bank column not found.")

    df = df.rename(columns={bank_col_name: "upi_remitter_banks"})

    # -----------------------------
    # Find Volume and TD columns
    # -----------------------------
    volume_col = None
    td_col = None
    for col in df.columns:
        c = str(col).lower().strip()

        if volume_col is None and "total" in c and "volume" in c:
            volume_col = col

        if td_col is None and c.startswith("td"):
            td_col = col

    if volume_col is None:
        raise ValueError("Total Volume column not found.")

    if td_col is None:
        raise ValueError("TD column not found.")

    df = df.rename(columns={
        volume_col: "total_volume_in_mn",
        td_col: "td_raw"
    })

    # Remove duplicate columns
    df = df.loc[:, ~df.columns.duplicated()]

    # -----------------------------
    # Standardize bank names
    # -----------------------------
    df["upi_remitter_banks"] = (
        df["upi_remitter_banks"]
        .astype(str)
        .str.strip()
        .str.lower()
        .replace({
            "state bank of india": "State Bank of India",
            "state bank of india ": "State Bank of India",
            "state bank of india.": "State Bank of India",
            "state bank of india ltd.": "State Bank of India",
            "hdfc bank ltd.": "HDFC Bank Ltd.",
            "hdfc bank ltd": "HDFC Bank Ltd.",
            "hdfc bank": "HDFC Bank Ltd.",
            "hdfc bank limited": "HDFC Bank Ltd.",
            "bank of baroda": "Bank of Baroda",
            "union bank of india": "Union Bank of India",
            "punjab national bank": "Punjab National Bank"
        })
    )

    # -----------------------------
    # Keep only required banks
    # -----------------------------
    df = df[df["upi_remitter_banks"].isin(target_banks)].copy()

    # -----------------------------
    # Clean Total Volume
    # -----------------------------
    df["total_volume_in_mn"] = (
        df["total_volume_in_mn"]
        .astype(str)
        .str.replace(",", "", regex=False)
        .astype(float)
    )

    # -----------------------------
    # Convert TD values
    # -----------------------------
    def convert_td(x):

        if pd.isna(x):
            return np.nan

        x = str(x).strip()

        if "%" in x:
            return float(x.replace("%", ""))

        val = float(x)

        # 2025 decimal format
        if val <= 0.05:
            return val * 100

        return val

    df["td_pct"] = df["td_raw"].apply(convert_td)

    # -----------------------------
    # Final output
    # -----------------------------
    df["month"] = month_label

    return df[
        [
            "month",
            "upi_remitter_banks",
            "total_volume_in_mn",
            "td_pct"
        ]
    ]
```

```python
import pandas as pd
import numpy as np
import os

def clean_upi_month(file_path, month_label, target_banks):

    # -----------------------------
    # Read file
    # -----------------------------
    if file_path.lower().endswith(".csv"):
        raw = pd.read_csv(file_path, header=None, dtype=str)
    else:
        raw = pd.read_excel(file_path, header=None, dtype=str)

    # -----------------------------
    # Detect header row automatically
    # -----------------------------
    header_row = None

    for i in range(len(raw)):

        row = (
            raw.iloc[i]
            .fillna("")
            .astype(str)
            .str.strip()
            .str.lower()
            .tolist()
        )

        if (
            any("remitter" in x for x in row)
            and any("volume" in x for x in row)
            and any("td" in x for x in row)
        ):
            header_row = i
            break

    if header_row is None:
        raise ValueError(f"Header row not found in {os.path.basename(file_path)}")

    # -----------------------------
    # Read again using detected header
    # -----------------------------
    if file_path.lower().endswith(".csv"):
        df = pd.read_csv(file_path, header=header_row)
    else:
        df = pd.read_excel(file_path, header=header_row)

    # -----------------------------
    # Standardize column names
    # -----------------------------
    df.columns = (
        df.columns.astype(str)
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
    )

    # Remove duplicate columns
    df = df.loc[:, ~df.columns.duplicated()]

    # -----------------------------
    # Find required columns
    # -----------------------------
    bank_col = None
    volume_col = None
    td_col = None

    for col in df.columns:

        c = str(col).lower()

        if bank_col is None and "remitter" in c:
            bank_col = col

        if volume_col is None and "volume" in c:
            volume_col = col

        if td_col is None and c.startswith("td"):
            td_col = col

    if bank_col is None:
        raise ValueError("Bank column not found.")

    if volume_col is None:
        raise ValueError("Volume column not found.")

    if td_col is None:
        raise ValueError("TD column not found.")

    df = df.rename(columns={
        bank_col: "upi_remitter_banks",
        volume_col: "total_volume_in_mn",
        td_col: "td_raw"
    })

    # -----------------------------
    # Standardize bank names
    # -----------------------------
    bank_mapping = {
        "state bank of india": "State Bank of India",
        "hdfc bank ltd.": "HDFC Bank Ltd.",
        "hdfc bank ltd": "HDFC Bank Ltd.",
        "hdfc bank limited": "HDFC Bank Ltd.",
        "hdfc bank": "HDFC Bank Ltd.",
        "bank of baroda": "Bank of Baroda",
        "union bank of india": "Union Bank of India",
        "punjab national bank": "Punjab National Bank"
    }

    df["upi_remitter_banks"] = (
        df["upi_remitter_banks"]
        .astype(str)
        .str.strip()
        .str.lower()
        .replace(bank_mapping)
    )

    # -----------------------------
    # Keep only target banks
    # -----------------------------
    df = df[df["upi_remitter_banks"].isin(target_banks)].copy()

    # -----------------------------
    # Clean Total Volume
    # -----------------------------
    df["total_volume_in_mn"] = (
        df["total_volume_in_mn"]
        .astype(str)
        .str.replace(",", "", regex=False)
        .astype(float)
    )

    # -----------------------------
    # Convert TD values
    # -----------------------------
    def convert_td(x):

        if pd.isna(x):
            return np.nan

        x = str(x).strip()

        if "%" in x:
            return float(x.replace("%", ""))

        x = float(x)

        if x <= 0.05:
            x *= 100

        return x

    df["td_pct"] = df["td_raw"].apply(convert_td)

    # -----------------------------
    # Final DataFrame
    # -----------------------------
    df["month"] = month_label

    return df[
        [
            "month",
            "upi_remitter_banks",
            "total_volume_in_mn",
            "td_pct"
        ]
    ]
```
